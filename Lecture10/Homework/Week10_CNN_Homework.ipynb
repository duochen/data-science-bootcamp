## 💡 Before You Start

A computer does not see an image the same way a person does.

An image is stored as numbers.

For a grayscale image:

- Each **pixel** contains one brightness value
- Dark pixels have smaller values
- Bright pixels have larger values

A color image usually has three channels:

- Red
- Green
- Blue

A CNN learns visual patterns by using small filters that move across the image.

Typical CNN workflow:

```text
Image
  ↓
Convolution
  ↓
Feature Maps
  ↓
Pooling
  ↓
More Convolution / Pooling
  ↓
Classification Layers
  ↓
Prediction
```

# 4️⃣ Understand Pixels and Tensors

Right now, each image has shape:

```text
(28, 28)
```

That means:

- 28 pixels high
- 28 pixels wide

A CNN expects an explicit channel dimension.

For grayscale images, we use:

```text
(28, 28, 1)
```

The final `1` means there is one grayscale channel.


````python
print("Original image shape:", X_train[0].shape)
print("Minimum pixel value:", X_train[0].min())
print("Maximum pixel value:", X_train[0].max())
````

# 5️⃣ Prepare the Images

Pixel values are currently between `0` and `255`.

We will scale them to the range `0` to `1`.

We will also add the grayscale channel dimension.

### Why scale?

Neural networks usually train more easily when input values are on a smaller, consistent scale.


````python
# Convert pixel values from integers to decimal numbers.
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Add a channel dimension for grayscale images.
X_train = np.expand_dims(X_train, axis=-1)
X_test = np.expand_dims(X_test, axis=-1)

print("Prepared training shape:", X_train.shape)
print("Prepared test shape:", X_test.shape)
print("New minimum pixel value:", X_train.min())
print("New maximum pixel value:", X_train.max())
````

### Option C — One More Epoch

### Important

Keep the rest of the setup the same so your comparison is fair.


````python
# TODO: Build and train your experimental CNN.

experiment_model = keras.Sequential([
    keras.Input(shape=(28, 28, 1)),

    # CHANGE ONE THING BELOW.
    layers.Conv2D(16, kernel_size=(3, 3), activation="relu"),
    layers.MaxPooling2D(pool_size=(2, 2)),

    layers.Conv2D(32, kernel_size=(3, 3), activation="relu"),
    layers.MaxPooling2D(pool_size=(2, 2)),

    layers.Flatten(),
    layers.Dense(32, activation="relu"),
    layers.Dense(10, activation="softmax")
])

experiment_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

experiment_history = experiment_model.fit(
    X_train,
    y_train,
    validation_split=0.1,
    epochs=3,
    batch_size=64,
    verbose=1
)

experiment_loss, experiment_accuracy = experiment_model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Experiment test accuracy:", round(experiment_accuracy, 4))
````
