## 1. The basic idea of an artificial neuron

A biological neuron receives signals from other neurons. An **artificial neuron** is a simplified mathematical idea inspired by this.

A neuron receives:

1. **Inputs** — the information entering the neuron.
2. **Weights** — how important each input is.
3. **Bias** — an adjustable value that shifts the neuron's decision.
4. **Activation function** — decides how the neuron transforms its result.

For one neuron:

\[
z = x_1w_1 + x_2w_2 + x_3w_3 + b
\]

Then we apply an activation function:

\[
output = activation(z)
\]


```python
import numpy as np

# Three example inputs.
# Imagine these are three measurements about a student:
# x1 = hours studied
# x2 = homework completion
# x3 = class participation
inputs = np.array([0.8, 0.6, 0.9])

# Each weight tells us how strongly that input affects the neuron.
weights = np.array([0.7, 0.4, 0.5])

# Bias lets the neuron shift its output up or down.
bias = -0.5

print("Inputs:", inputs)
print("Weights:", weights)
print("Bias:", bias)
```

### Calculate the weighted sum

The neuron multiplies each input by its weight, adds the results, and then adds the bias.

`np.dot(inputs, weights)` performs the multiplication and addition for us.


```python
# Calculate z = x1*w1 + x2*w2 + x3*w3 + bias
z = np.dot(inputs, weights) + bias

print("Weighted sum (z):", z)
```

### Check the calculation step by step

Try to predict the value before running the next cell.

\[
z = (0.8)(0.7) + (0.6)(0.4) + (0.9)(0.5) - 0.5
\]


```python
# Show each contribution separately.
contributions = inputs * weights

print("Input × weight contributions:", contributions)
print("Sum of contributions:", contributions.sum())
print("After adding bias:", contributions.sum() + bias)
```

## 2. Activation functions

Without an activation function, a neural network would be much less powerful.

Two common activation functions are:

### ReLU

\[
ReLU(z) = max(0, z)
\]

- Negative values become `0`.
- Positive values stay positive.
- ReLU is commonly used in **hidden layers**.

### Sigmoid

\[
sigmoid(z) = \frac{1}{1 + e^{-z}}
\]

- Converts any number into a value between `0` and `1`.
- It is useful when an output can be interpreted like a probability for a **binary** decision.


```python
def relu(z):
    # ReLU keeps positive values and changes negative values to 0.
    return np.maximum(0, z)

def sigmoid(z):
    # Sigmoid converts a number into the range 0 to 1.
    return 1 / (1 + np.exp(-z))

print("z =", z)
print("ReLU(z) =", relu(z))
print("Sigmoid(z) =", sigmoid(z))
```

## 5. Forward propagation through a tiny network

A neural network usually contains many neurons arranged in **layers**.

We will create:

- **Input layer:** 3 input values
- **Hidden layer:** 2 neurons
- **Output layer:** 1 neuron

**Forward propagation** means that data moves from the input layer → hidden layer → output layer.


```python
# We have 3 inputs and 2 hidden neurons.
# Therefore, the weight matrix has shape (3, 2).
hidden_weights = np.array([
    [0.6, -0.2],
    [0.3,  0.8],
    [0.5,  0.4]
])

# One bias value for each hidden neuron.
hidden_bias = np.array([0.1, -0.3])

# Step 1: Calculate the hidden layer's weighted sums.
hidden_z = np.dot(inputs, hidden_weights) + hidden_bias

# Step 2: Apply ReLU.
hidden_output = relu(hidden_z)

print("Hidden weighted sums:", hidden_z)
print("Hidden layer after ReLU:", hidden_output)
```

```python
# The hidden layer has 2 outputs, so the output neuron needs 2 weights.
output_weights = np.array([0.9, -0.4])
output_bias = 0.2

# Step 3: Send the hidden-layer outputs to the output neuron.
output_z = np.dot(hidden_output, output_weights) + output_bias

# Step 4: Apply sigmoid to get a value between 0 and 1.
prediction = sigmoid(output_z)

print("Output weighted sum:", output_z)
print("Final network output:", prediction)
```
