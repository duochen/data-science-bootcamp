## Lists

```python
scores = [88, 95, 76, 92, 81]  # Create a list that stores five scores.
print(scores)  # Display the whole list.
print(scores[0])  # Display the first score because Python starts counting at 0.
print(scores[-1])  # Display the last score in the list.
practice_scores = [90, 84, 93]  # Create a practice list with three scores.
practice_scores.append(88)  # Add one more score to the end of the list.
print(practice_scores[1])  # Display the second score.
print(len(practice_scores))  # Display the number of values in the list.
```

## Loops

```python
temperatures = [72, 75, 79, 81, 77]  # Create a list of daily temperatures.
for temperature in temperatures:  # Repeat the next line once for each temperature.
    print(temperature)  # Display the current temperature.
numbers = [1, 2, 3, 4, 5]  # Create a list containing the numbers 1 through 5.
for number in numbers:  # Repeat the next line once for each number.
    print(number * 2)  # Multiply the current number by 2 and display the result.
```

## Functions

```python
def calculate_average(first_score, second_score):  # Define a function that accepts two scores.
    average = (first_score + second_score) / 2  # Calculate the average of the two scores.
    return average  # Send the calculated result back.
result = calculate_average(88, 94)  # Call the function with two example scores.
print(result)  # Display the first calculated average.
second_result = calculate_average(90, 96)  # Call the same function with two different scores.
print(second_result)  # Display the second calculated average.
```

## Statistics

```python
average_math = data["math"].mean()  # Calculate the average math score.
average_science = data["science"].mean()  # Calculate the average science score.
highest_math = data["math"].max()  # Find the highest math score.
average_study_hours = data["study_hours"].mean()  # Calculate the average number of study hours.
highest_science = data["science"].max()  # Find the highest science score.
lowest_math = data["math"].min()  # Find the lowest math score.
print(average_math)  # Display the average math score.
print(average_science)  # Display the average science score.
print(highest_math)  # Display the highest math score.
print(average_study_hours)  # Display the average number of study hours.
print(highest_science)  # Display the highest science score.
print(lowest_math)  # Display the lowest math score.
```

## Student Who Studied Most

```python
most_hours = data["study_hours"].max()  # Find the largest number of study hours.
student_index = data["study_hours"].idxmax()  # Find the row index containing the largest study-hours value.
student_name = data.loc[student_index, "student"]  # Use the row index to find the student's name.
print(most_hours)  # Display the largest number of study hours.
print(student_name)  # Display the student who studied the most.
```

## New Columns
