## 1. What Is Google Colab?

Google Colab, short for **Google Colaboratory**, is a web-based environment for writing and running Python code.

You usually do not need to install Python on your own computer.

A Colab notebook is made of **cells**.

There are two main types of cells:

1. **Code cells** — contain Python code that you can run.
2. **Text cells** — contain explanations, headings, notes, and instructions.

Colab notebooks use the `.ipynb` file format, which is also used by Jupyter Notebook.

### Why are notebooks useful for data science?

A notebook can keep all of these together:

- explanations,
- Python code,
- data,
- results,
- charts,
- notes,
- experiments.

That makes notebooks useful for **reproducible data science**, because another person can read your work and rerun your code.

## 6. Lists

A **list** stores multiple values.

Python starts counting positions at **0**.

```python
scores = [88, 95, 76, 92, 81]  # Create a list that stores five scores.
print(scores)  # Display the whole list of scores.
print(scores[0])  # Display the first score because index 0 means the first item.
print(scores[-1])  # Display the last score because index -1 means the final item.
```

## 7. Loops

A loop repeats an action.

```python
temperatures = [72, 75, 79, 81, 77]  # Create a list of daily temperatures.
for temperature in temperatures:  # Repeat the next indented line once for each temperature in the list.
    print(temperature)  # Display the current temperature during each loop repetition.
```

## 8. Functions

A function packages reusable logic into a named block of code.

```python
def calculate_average(first_score, second_score):  # Define a function that accepts two scores.
    average = (first_score + second_score) / 2  # Add the two scores and divide by 2 to calculate the average.
    return average  # Send the calculated average back to the place where the function was called.
result = calculate_average(88, 94)  # Call the function with two scores and store the returned result.
print(result)  # Display the calculated average.
```

## 10. What Is a Colab Runtime?

A **runtime** is the temporary computer session that runs your code.

If the runtime restarts:

- your variables disappear,
- files uploaded only to the temporary runtime may disappear,
- you may need to rerun earlier cells.

Important work should be saved to:

- your notebook,
- Google Drive,
- GitHub,
- or your own computer.

```python
import os  # Import Python's operating-system tools so we can inspect the current environment.
print(os.getcwd())  # Display the current working folder used by the notebook.
print(os.listdir("."))  # Display the files that currently exist in the working folder.
```

## 13. Read a CSV File with Pandas

A **CSV file** is a simple table where values are separated by commas.

Pandas reads a CSV file into a **DataFrame**.

A DataFrame is a table with rows and columns.

```python
data = pd.read_csv("student_scores.csv")  # Read the CSV file into a Pandas DataFrame named data.
print(data)  # Display the complete DataFrame as plain text output.
```

## 14. Inspect the Dataset

Data scientists usually inspect a dataset before analyzing it.

## 15. Select Columns

You can select one column from a DataFrame by using its column name.

```python
math_scores = data["math"]  # Select the math column and store it in a new variable.
science_scores = data["science"]  # Select the science column and store it in a new variable.
print(math_scores)  # Display the math scores.
print(science_scores)  # Display the science scores.
```

## 16. Simple Data Analysis

Pandas can calculate useful statistics quickly.

```python
average_math = data["math"].mean()  # Calculate the average value in the math column.
average_science = data["science"].mean()  # Calculate the average value in the science column.
highest_math = data["math"].max()  # Find the highest value in the math column.
print(average_math)  # Display the average math score.
print(average_science)  # Display the average science score.
print(highest_math)  # Display the highest math score.
```

### Exercise 7

Use Python to answer:

- Which subject has the higher average?
- Which student studied the most hours?

```python
print(data["study_hours"].max())  # Display the largest number of study hours in the dataset.
print(data.loc[data["study_hours"].idxmax(), "student"])  # Find and display the student whose study-hours value is the largest.
```

## 17. Create a New Column

Creating a value from existing columns is a simple example of **feature engineering**.

```python
data["average_score"] = (data["math"] + data["science"]) / 2  # Create a new column by averaging each student's math and science scores.
print(data)  # Display the updated DataFrame so we can see the new column.
```

## 18. Create a Simple Chart

Charts help us see patterns in data.

```python
plt.bar(data["student"], data["average_score"])  # Create a bar chart using student names and average scores.
plt.title("Average Score by Student")  # Add a clear title to the chart.
plt.xlabel("Student")  # Label the horizontal axis.
plt.ylabel("Average Score")  # Label the vertical axis.
plt.xticks(rotation=45)  # Rotate student names so they are easier to read.
plt.tight_layout()  # Adjust spacing so labels fit inside the chart area.
plt.show()  # Display the finished chart.
```

### Exercise 8

Change the chart so it shows **study hours by student**.
