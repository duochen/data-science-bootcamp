## 1. Getting Started

A **NumPy array** is a container that holds many numbers at once. It is much faster and uses less memory than a regular Python list, which is why data scientists use it everywhere.

```python
# Import the NumPy library and give it the short nickname 'np' so we type less
import numpy as np
```

## 2. Creating Arrays

There are many ways to make a NumPy array — from a list, from a range, or filled with zeros, ones, or random numbers.

```python
# Make a Python list of integers
nums_list = [10, 12, 14, 16, 20]
# Convert the Python list into a NumPy array
nums_array = np.array(nums_list)
# Print the array so we can see it
print(nums_array)
# Check the type — it should be numpy.ndarray (n-dimensional array)
print(type(nums_array))
```

```python
# A 2D array (a matrix) is a list of lists — each inner list is one row
row1 = [10, 12, 13]   # first row
row2 = [45, 32, 16]   # second row
row3 = [50, 60, 70]   # third row
# Combine the rows into a 2D NumPy array
nums_2d = np.array([row1, row2, row3])
# Print the whole 2D array
print(nums_2d)
# .shape tells us (number_of_rows, number_of_columns)
print(nums_2d.shape)
```

```python
# NumPy also supports strings, not just numbers
colors = np.array(["Red", "Green", "Orange"])
# Show the array of strings
print(colors)
# .dtype shows what kind of data is stored ('<U6' means Unicode text, up to 6 chars)
print(colors.dtype)
```

```python
# np.arange creates a range of numbers: start (included), stop (NOT included)
# This gives us 5, 6, 7, 8, 9, 10 — the number 11 is NOT included
nums_arr = np.arange(5, 11)
print(nums_arr)
# We can also give a step size — here we jump by 2 each time
nums_arr2 = np.arange(5, 12, 2)
print(nums_arr2)
```

```python
# np.ones creates an array filled with the number 1.0
ones_1d = np.ones(6)
print(ones_1d)
# Pass a tuple (rows, columns) to make a 2D array of ones
ones_2d = np.ones((3, 4))
print(ones_2d)
```

```python
# np.zeros works the same way but fills with 0.0
zeros_1d = np.zeros(6)
print(zeros_1d)
# 2D version: 3 rows and 4 columns of zeros
zeros_2d = np.zeros((3, 4))
print(zeros_2d)
```

```python
# np.eye creates an 'identity matrix' — 1's on the diagonal, 0's everywhere else
# It is used a lot in linear algebra (advanced math)
identity = np.eye(5)
print(identity)
```

```python
# np.random.rand creates random decimals between 0 and 1
# Here we make a 3x4 grid of random decimals
uniform_random = np.random.rand(3, 4)
print(uniform_random)
# np.random.randint creates random WHOLE numbers between low (included) and high (excluded)
# The size argument controls the shape — here 5 random integers between 10 and 49
integer_random = np.random.randint(10, 50, size=5)
print(integer_random)
```

## 3. Array Properties and Iteration

Every array has a **shape** (its size), a **number of dimensions** (`ndim`), and we can **loop through** its values.

```python
# Create a 1D array
my_array = np.array([10, 12, 14, 16, 20, 25])
# .ndim = number of dimensions — 1 for a flat list
print(my_array.ndim)
# .shape = size along each dimension — (6,) means 6 items in one dimension
print(my_array.shape)
```

```python
# Create a 2D array of random integers, 4 rows and 5 columns
integer_random = np.random.randint(1, 11, size=(4, 5))
print(integer_random)
# For a 2D array, .ndim is 2
print(integer_random.ndim)
# .shape gives (rows, columns)
print(integer_random.shape)
```

```python
# Loop through a 2D array to visit every element
# The outer loop goes through each row
for row in integer_random:
    # The inner loop goes through each value inside that row
    for value in row:
        # Print each value one by one
        print(value)
```

## 4. Adding and Removing Items

Use `np.append` to add and `np.delete` to remove. Both return a **new** array — the original is unchanged.

```python
# Start with an array of colors
colors = np.array(["Red", "Green", "Orange"])
# np.append returns a NEW array with an extra item at the end
extended = np.append(colors, "Yellow")
print(extended)
# We can also append several items at once — pass them in a list
extended2 = np.append(colors, ["Yellow", "Pink"])
print(extended2)
```

```python
# Start with a 3x3 grid of zeros
zeros_array = np.zeros((3, 3))
# Append a new row [1, 2, 3] to the bottom
# axis=0 means 'add along the row direction' (make the array taller)
extended_rows = np.append(zeros_array, [[1, 2, 3]], axis=0)
print(extended_rows)
# axis=1 means 'add along the column direction' (make the array wider)
extended_cols = np.append(zeros_array, [[1], [2], [3]], axis=1)
print(extended_cols)
```

```python
# Delete a single item by its index (position, starting from 0)
colors = np.array(["Red", "Green", "Orange"])
# Remove the item at index 1 (which is 'Green')
updated = np.delete(colors, 1)
print(updated)
```

```python
# Delete rows or columns from a 2D array
integer_random = np.random.randint(1, 11, size=(4, 5))
print(integer_random)
# axis=0 removes a ROW — here row at index 1
without_row = np.delete(integer_random, 1, axis=0)
print(without_row)
# axis=1 removes a COLUMN — here column at index 1
without_col = np.delete(integer_random, 1, axis=1)
print(without_col)
```

## 5. Sorting

```python
# Create 10 random integers between 1 and 19
my_array = np.random.randint(1, 20, 10)
print(my_array)
# np.sort returns a NEW array sorted from smallest to largest (ascending)
sorted_array = np.sort(my_array)
print(sorted_array)
```

```python
# To sort from largest to smallest (descending):
# first sort ascending, then flip the order using np.flipud
descending = np.flipud(sorted_array)
print(descending)
```

```python
# Text arrays are sorted alphabetically
colors = np.array(["Red", "Green", "Blue", "Yellow"])
print(np.sort(colors))
```

## 6. Reshaping

Reshaping changes the shape (rows and columns) of an array **without changing its data**. The total number of elements must stay the same.

```python
# Create a 1D array with 10 random numbers
one_d = np.random.randint(1, 20, 10)
print(one_d)
# Reshape into 2 rows and 5 columns (2 * 5 = 10, so it fits perfectly)
two_d = one_d.reshape(2, 5)
print(two_d)
```

```python
# Flatten a 2D array back into 1D
two_d = np.random.randint(1, 20, size=(4, 6))
print(two_d)
# Passing -1 tells NumPy to figure out the size automatically
flat = two_d.reshape(-1)
print(flat)
```

## 7. Indexing and Slicing

We can grab a single item or a **slice** (a range of items). Positions start from 0.

```python
# An array with numbers 1 through 10
s = np.arange(1, 11)
print(s)
# Get the item at index 1 (the second item)
print(s[1])
# Slice: from index 1 up to (but not including) index 9
print(s[1:9])
# First 5 items
print(s[:5])
# Everything from index 5 to the end
print(s[5:])
```

```python
# Slicing works on 2D arrays too — use [rows, columns]
nums_2d = np.array([[10, 12, 13],
                    [45, 32, 16],
                    [50, 60, 70]])
# First two rows, all columns
print(nums_2d[:2, :])
# All rows, first two columns
print(nums_2d[:, :2])
# Rows from index 1 onwards AND columns from index 1 onwards
print(nums_2d[1:, 1:])
```

## 8. Broadcasting

Broadcasting lets NumPy do math between arrays of **different shapes** automatically — no need to write loops.

```python
# Make a 3x4 array
array1 = np.random.randint(1, 20, size=(3, 4))
print(array1)
# Add 10 to EVERY element in one line — NumPy 'broadcasts' 10 to each cell
result = array1 + 10
print(result)
```

```python
# We can also broadcast a smaller array across a larger one
array1 = np.random.randint(1, 20, size=(3, 4))
print(array1)
# This 1D array of 4 values gets added to each row of the 3x4 array
row_to_add = np.array([5, 10, 20, 25])
result = array1 + row_to_add
print(result)
```

## 9. Copy vs View

- A **copy** is independent — changes to it do NOT affect the original.
- A **view** shares data with the original — changing one DOES change the other.

```python
# COPY: independent from the original
array1 = np.array([10, 12, 14, 16, 18, 20])
# .copy() creates a brand new array with its own data
array2 = array1.copy()
# Change one element in array2
array2[1] = 99
# array1 is UNCHANGED
print(array1)
print(array2)
```

```python
# VIEW: shares data with the original
array1 = np.array([10, 12, 14, 16, 18, 20])
# .view() looks like a new array but points to the same data underneath
array2 = array1.view()
# Change array2
array2[1] = 99
# array1 CHANGES too — they share memory!
print(array1)
print(array2)
```

## 10. Statistics with NumPy

NumPy has built-in functions for summarizing data — the foundation of every data-science project.

```python
# Imagine these numbers are test scores
scores = np.array([2, 4, 8, 10, 12])
# Mean = the average of all values
print("mean:", np.mean(scores))
# Median = the middle value when the numbers are sorted
print("median:", np.median(scores))
# Min = the smallest value
print("min:", np.amin(scores))
# Max = the largest value
print("max:", np.amax(scores))
# Std (standard deviation) = how 'spread out' the values are
print("std:", np.std(scores))
```

```python
# For 2D arrays, we can compute stats for each row or each column using 'axis'
my_array = np.random.randint(1, 20, size=(3, 4))
print(my_array)
# axis=1 → one value per ROW
print("row means:", np.mean(my_array, axis=1))
# axis=0 → one value per COLUMN
print("column means:", np.mean(my_array, axis=0))
```

## 11. Unique Values and Counts

`np.unique` finds the different values in an array — and can also count how many times each appears.

```python
# An array with some repeated numbers
my_array = np.array([5, 8, 7, 5, 9, 3, 7, 7, 1, 1, 8, 4, 6, 9, 7, 3])
# return_counts=True returns two arrays: the unique values, AND how often each appears
unique_items, counts = np.unique(my_array, return_counts=True)
print("unique values:", unique_items)
print("counts:", counts)
```

## 13. Plotting with Matplotlib

**Matplotlib** turns arrays into charts. This is where NumPy really shines for data science.

```python
# Import matplotlib's pyplot module (nickname 'plt' is standard)
import matplotlib.pyplot as plt
# Create 20 evenly spaced values between 0 and 20 for the x-axis
x_vals = np.linspace(0, 20, 20)
# Compute the square root of each x value for the y-axis
y_vals = np.sqrt(x_vals)
# Draw the line: x_vals along the bottom, y_vals going up
plt.plot(x_vals, y_vals)
# Actually display the chart
plt.show()
```

```python
# Plot TWO lines on the same chart with labels and a legend
x_vals = np.linspace(0, 20, 20)
# First line: square roots (grows slowly)
y_vals = np.sqrt(x_vals)
# Second line: cubes (grows very fast)
y2_vals = x_vals ** 3
# Label the axes so the reader knows what they mean
plt.xlabel('X Values')
plt.ylabel('Y Values')
# Give the whole chart a title
plt.title('Square Roots vs Cubes')
# 'r' = red line, 'b' = blue line; label= adds a name to the legend
plt.plot(x_vals, y_vals, 'r', label='Square Root')
plt.plot(x_vals, y2_vals, 'b', label='Cube')
# Show a legend at the top center of the chart
plt.legend(loc='upper center')
# Display the finished chart
plt.show()
```

```python
# We can also display a 2D array as a colored grid (a 'heatmap')
my_array = np.random.randint(1, 20, size=(4, 4))
print(my_array)
# imshow paints each number as a colored square
plt.imshow(my_array)
# Add a color bar so we know which color means which number
plt.colorbar()
plt.show()
```
