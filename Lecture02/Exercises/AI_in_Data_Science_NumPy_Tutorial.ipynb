## 1. What is NumPy?

NumPy stands for **Numerical Python**. Its main object is the `ndarray`, an efficient container for values that usually share one data type.

Data science and machine-learning libraries use arrays because datasets are often rectangular: rows represent observations, while columns represent features.

### Python list versus NumPy array

A Python list is flexible. A NumPy array is designed for fast numerical calculations. The `+` operator therefore behaves differently.

```python
import numpy as np  # Import NumPy and use the standard short name np.
python_scores = [70, 80, 90]  # Create a regular Python list of three scores.
numpy_scores = np.array([70, 80, 90])  # Convert the scores into a NumPy array.
print(python_scores + python_scores)  # Join two Python lists instead of adding their values.
print(numpy_scores + numpy_scores)  # Add matching NumPy array values element by element.
```

> **Check:** Why did the first result contain six values while the second contained three?  
> **Answer:** List `+` means concatenation; array `+` means element-wise addition.

## 2. Creating arrays

- `np.array(...)` converts existing data.
- `np.arange(start, stop, step)` creates evenly stepped values; `stop` is excluded.
- `np.linspace(start, stop, count)` creates a requested number of evenly spaced values; both endpoints are included.
- `np.zeros(...)` and `np.ones(...)` create arrays with known starting values.

```python
levels = np.arange(1, 11)  # Create level numbers from 1 through 10 because 11 is excluded.
checkpoints = np.linspace(0, 100, 5)  # Create five equally spaced checkpoint percentages from 0 to 100.
empty_scores = np.zeros(4, dtype=int)  # Create four integer score slots initialized to zero.
bonus_flags = np.ones(4, dtype=bool)  # Create four Boolean flags initialized to True.
print(levels)  # Display the level numbers.
print(checkpoints)  # Display the checkpoint percentages.
print(empty_scores)  # Display the empty integer score slots.
print(bonus_flags)  # Display the Boolean flags.
```

## 3. Understanding array structure

The most important array properties are:

- `ndim`: number of dimensions
- `shape`: length along each dimension
- `size`: total number of values
- `dtype`: data type stored in the array

In the table below, each **row** is a student and each **column** is a quiz.

```python
student_scores = np.array([[82, 88, 91, 86], [75, 79, 84, 88], [93, 90, 95, 97]])  # Create a 3-student by 4-quiz score array.
print(student_scores)  # Display the complete two-dimensional array.
print("Dimensions:", student_scores.ndim)  # Display the number of array dimensions.
print("Shape:", student_scores.shape)  # Display the number of rows and columns.
print("Size:", student_scores.size)  # Display the total number of stored scores.
print("Data type:", student_scores.dtype)  # Display the type used to store each score.
```

> **Predict:** What shape would an array need for 20 students taking 5 quizzes?  
> **Answer:** `(20, 5)` because rows are students and columns are quizzes.

## 4. Indexing and slicing

Python starts counting at index `0`. For a 2D array, use `[row, column]`.

A slice uses `start:stop`; the start is included and the stop is excluded. A colon by itself means “all values along this dimension.”

```python
first_student = student_scores[0]  # Select every quiz score in the first row.
second_quiz = student_scores[:, 1]  # Select the second column for every student.
one_score = student_scores[2, 3]  # Select the fourth quiz score for the third student.
first_two_students_first_three_quizzes = student_scores[:2, :3]  # Select the first two rows and first three columns.
print(first_student)  # Display all scores for the first student.
print(second_quiz)  # Display Quiz 2 scores for all students.
print(one_score)  # Display the selected single score.
print(first_two_students_first_three_quizzes)  # Display the selected rectangular section.
```

### Views and copies: an important safety rule

Many NumPy slices are **views**, so changing the slice may change the original array. Use `.copy()` when you need an independent array.

```python
practice_scores = student_scores.copy()  # Make an independent copy so the original classroom data stays unchanged.
score_view = practice_scores[0, :2]  # Create a view of the first student's first two scores.
score_view[0] = 100  # Change the first value in the view to demonstrate shared data.
print(practice_scores[0])  # Show that the original copied array also changed through the view.
safe_copy = practice_scores[1, :2].copy()  # Make an independent copy of two values.
safe_copy[0] = 100  # Change only the independent copy.
print(practice_scores[1])  # Show that the source row did not change.
```

## 5. Boolean filtering

A comparison creates a Boolean mask containing `True` and `False`. Applying the mask keeps only values where the mask is `True`.

```python
high_score_mask = student_scores >= 90  # Test every score and mark values of 90 or higher as True.
high_scores = student_scores[high_score_mask]  # Keep only scores whose mask value is True.
print(high_score_mask)  # Display the Boolean mask with the same shape as the data.
print(high_scores)  # Display the selected high scores as a one-dimensional array.
print("Number of high scores:", np.sum(high_score_mask))  # Count True values because True acts like 1.
```

### Combining conditions

Use `&` for **and**, `|` for **or**, and `~` for **not**. Put each comparison inside parentheses.

```python
middle_scores = student_scores[(student_scores >= 80) & (student_scores < 90)]  # Select scores from 80 through 89 using two conditions.
print(middle_scores)  # Display the scores that satisfy both conditions.
```

## 6. Reshaping arrays

Reshaping changes how values are arranged without changing how many values exist. The new shape must have the same total size.

```python
game_scores = np.arange(10, 70, 10)  # Create six game scores from 10 through 60.
score_board = game_scores.reshape(2, 3)  # Rearrange the six values into two players and three rounds.
flat_scores = score_board.reshape(-1)  # Flatten the array and let NumPy calculate the required length.
print(score_board)  # Display the two-dimensional score board.
print(flat_scores)  # Display the same values in one dimension.
```

## 7. Vectorized operations

Vectorization applies an operation to an entire array without writing a Python loop. This makes code shorter and usually faster.

```python
raw_scores = np.array([68, 75, 84, 91])  # Create four raw student scores.
curved_scores = np.minimum(raw_scores + 5, 100)  # Add five points to every score and cap results at 100.
percent_change = (curved_scores - raw_scores) / raw_scores * 100  # Calculate each score's percentage increase.
print(curved_scores)  # Display the curved scores.
print(np.round(percent_change, 1))  # Display percentage changes rounded to one decimal place.
```

## 8. Broadcasting

Broadcasting lets NumPy combine arrays with compatible shapes. Here, one bonus per round is added to every player's score in that round.

The score matrix has shape `(3, 4)`. The bonus array has shape `(4,)`, so NumPy aligns it with the four columns.

```python
player_scores = np.array([[120, 135, 128, 142], [110, 145, 138, 150], [130, 125, 140, 148]])  # Store three players' scores across four rounds.
round_bonuses = np.array([0, 5, 10, 15])  # Store one bonus for each of the four rounds.
adjusted_scores = player_scores + round_bonuses  # Broadcast the four bonuses across all three player rows.
print(adjusted_scores)  # Display the adjusted scores.
```

> **Predict:** Why would adding an array with shape `(3,)` fail here?  
> **Answer:** NumPy compares dimensions from right to left. The final dimensions `4` and `3` are incompatible.

## 9. Aggregation and the meaning of `axis`

An aggregation summarizes many values.

- No `axis`: summarize the whole array.
- `axis=0`: collapse the rows and calculate one result per column.
- `axis=1`: collapse the columns and calculate one result per row.

For our student-score table, `axis=0` gives a result for each quiz and `axis=1` gives a result for each student.

```python
overall_mean = np.mean(student_scores)  # Calculate one mean using every score.
quiz_means = np.mean(student_scores, axis=0)  # Calculate one mean for each quiz column.
student_means = np.mean(student_scores, axis=1)  # Calculate one mean for each student row.
score_median = np.median(student_scores)  # Find the middle score across the entire dataset.
score_spread = np.std(student_scores)  # Measure how spread out all scores are using population standard deviation.
lowest_score = np.min(student_scores)  # Find the lowest score in the entire array.
highest_score = np.max(student_scores)  # Find the highest score in the entire array.
print("Overall mean:", np.round(overall_mean, 2))  # Display the overall mean rounded to two decimals.
print("Quiz means:", np.round(quiz_means, 2))  # Display one mean for each quiz.
print("Student means:", np.round(student_means, 2))  # Display one mean for each student.
print("Median:", score_median)  # Display the median score.
print("Standard deviation:", np.round(score_spread, 2))  # Display the population standard deviation.
print("Range:", lowest_score, "to", highest_score)  # Display the minimum and maximum scores.
```

### Finding positions with `argmin` and `argmax`

For a 1D array, `argmax()` returns the position of the largest value. We can use that position to retrieve a matching name.

```python
student_names = np.array(["Ava", "Ben", "Carlos"])  # Store names in the same row order as the score table.
best_student_index = np.argmax(student_means)  # Find the row index with the highest student mean.
best_student_name = student_names[best_student_index]  # Use the index to select the matching student name.
print("Highest average:", best_student_name, np.round(student_means[best_student_index], 2))  # Display the top student's name and average.
```

## 10. Reproducible random data

Simulated data is useful for testing an analysis before real data is available. A random-number generator with a fixed seed produces the same values every run, making an experiment reproducible.

Use the modern `default_rng()` interface rather than changing NumPy's global random state.

```python
rng = np.random.default_rng(42)  # Create a modern random generator with a reproducible seed.
simulated_game_scores = rng.integers(50, 101, size=(5, 4))  # Generate scores from 50 through 100 for five players and four rounds.
print(simulated_game_scores)  # Display the reproducible simulated scores.
```

## 11. Guided mini-project: Game Tournament Analyzer

### Scenario

Four players complete five rounds of a strategy game. Rows are players and columns are rounds. Analyze performance and identify patterns.

Before running each cell, predict the output's **shape** and what it will mean.

```python
players = np.array(["Maya", "Noah", "Liam", "Zoe"])  # Store player names in the same order as the score rows.
tournament_scores = np.array([[82, 91, 88, 95, 90], [76, 84, 89, 87, 92], [90, 86, 94, 91, 96], [85, 88, 82, 90, 93]])  # Store four players' scores across five rounds.
player_averages = np.mean(tournament_scores, axis=1)  # Calculate one average for each player row.
round_averages = np.mean(tournament_scores, axis=0)  # Calculate one average for each round column.
consistency = np.std(tournament_scores, axis=1)  # Calculate each player's score variation across rounds.
winner_index = np.argmax(player_averages)  # Find the row index of the player with the highest average.
most_consistent_index = np.argmin(consistency)  # Find the row index of the player with the smallest standard deviation.
improvement = tournament_scores[:, -1] - tournament_scores[:, 0]  # Subtract each player's first-round score from the final-round score.
improved_players = players[improvement > 0]  # Select names of players whose final score exceeded their first score.
print("Player averages:", np.round(player_averages, 1))  # Display each player's average rounded to one decimal.
print("Round averages:", np.round(round_averages, 1))  # Display each round's average rounded to one decimal.
print("Tournament winner:", players[winner_index])  # Display the player with the highest average.
print("Most consistent:", players[most_consistent_index])  # Display the player with the smallest variation.
print("Improvement:", improvement)  # Display each player's change from first to final round.
print("Players who improved:", improved_players)  # Display the names of players with positive improvement.
```

## 12. Common mistakes and debugging checklist

1. **Wrong shape:** Print `.shape` before combining arrays.
2. **Off-by-one indexing:** Remember that index `0` is the first position.
3. **Wrong axis:** Say what each row and column represents before choosing an axis.
4. **Unexpected source changes:** A slice may be a view; use `.copy()` when independence matters.
5. **Incompatible broadcasting:** Compare shapes from right to left.
6. **Unreproducible random results:** Create a generator with a fixed seed.
7. **A calculation without interpretation:** Always explain what a number means in the real scenario.
