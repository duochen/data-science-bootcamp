```python
import numpy as np  # Import NumPy and use the standard short name np.
```

## 1. Create and inspect an array

Create a 2D NumPy array named `scores` from the provided data. Then display its shape, number of dimensions, total size, and data type.

Rows represent students; columns represent four quizzes.

**Hint:** Use `np.array(...)`, then the properties `.shape`, `.ndim`, `.size`, and `.dtype`.

```python
score_data = [[78, 84, 91, 87], [92, 89, 94, 96], [70, 75, 80, 85], [88, 90, 86, 93], [81, 79, 85, 88]]  # Store the provided score data as nested Python lists.
scores = np.array(score_data)  # Convert the nested lists into a two-dimensional NumPy array.
print(scores)  # Display the complete score array.
print("Shape:", scores.shape)  # Display the number of student rows and quiz columns.
print("Dimensions:", scores.ndim)  # Display the number of array dimensions.
print("Size:", scores.size)  # Display the total number of scores.
print("Data type:", scores.dtype)  # Display the data type used for each score.
```

## 2. Index and slice

Using `scores`, display:

1. The third student's complete row.
2. Every student's Quiz 2 score.
3. The first three students' scores on the first two quizzes.

**Hint:** Use `[row, column]`; a colon selects a range or all entries on an axis.

```python
third_student = scores[2]  # Select every quiz score from the third student row.
quiz_two = scores[:, 1]  # Select the second quiz column for every student.
three_by_two = scores[:3, :2]  # Select the first three rows and first two columns.
print(third_student)  # Display the third student's scores.
print(quiz_two)  # Display every student's Quiz 2 score.
print(three_by_two)  # Display the requested three-row by two-column section.
```

## 3. Filter with conditions

Find and display:

1. Every score of 90 or higher.
2. The number of scores below 80.
3. Every score from 80 through 89, inclusive.

**Hint:** Comparisons produce Boolean masks. Combine conditions with `&`, and put each comparison in parentheses.

```python
scores_at_least_90 = scores[scores >= 90]  # Keep only scores that are 90 or higher.
number_below_80 = np.sum(scores < 80)  # Count True values in the below-80 Boolean mask.
scores_in_80s = scores[(scores >= 80) & (scores <= 89)]  # Keep scores from 80 through 89 using two conditions.
print(scores_at_least_90)  # Display all scores of 90 or higher.
print(number_below_80)  # Display the number of scores below 80.
print(scores_in_80s)  # Display all scores from 80 through 89.
```

## 4. Reshape game data

The array below contains 12 scores recorded in order. Reshape it into three players and four rounds. Then flatten it back into one dimension.

**Hint:** Use `.reshape(3, 4)` and `.reshape(-1)`.

```python
recorded_scores = np.array([55, 62, 71, 80, 60, 68, 74, 85, 58, 65, 79, 90])  # Store 12 game scores in recorded order.
game_matrix = recorded_scores.reshape(3, 4)  # Reshape the values into three players and four rounds.
flattened_scores = game_matrix.reshape(-1)  # Flatten the matrix back into one dimension.
print(game_matrix)  # Display the three-by-four game matrix.
print(flattened_scores)  # Display the flattened scores.
```

## 5. Vectorization and broadcasting

Award every player 10% extra points, rounded to one decimal place. Then add the round bonuses `[0, 2, 4, 6]` to the corresponding columns.

Do not write a loop.

**Hint:** Multiply the entire matrix by `1.10`. A four-value bonus array can broadcast across four columns.

```python
round_bonuses = np.array([0, 2, 4, 6])  # Store one bonus for each round column.
increased_scores = np.round(game_matrix * 1.10, 1)  # Increase every score by 10 percent and round to one decimal place.
adjusted_game_scores = increased_scores + round_bonuses  # Broadcast the round bonuses across all player rows.
print(adjusted_game_scores)  # Display the final adjusted scores.
```

## 6. Analyze student performance

Using the original `scores` array, calculate and display:

1. The average for each student.
2. The average for each quiz.
3. Each student's highest score.
4. Each student's standard deviation.
5. The name of the student with the highest average.

**Hint:** Rows are students. Which axis produces one result per row? Use `np.argmax(...)` to locate the highest average.

```python
student_names = np.array(["Aiden", "Bella", "Chloe", "Diego", "Emma"])  # Store student names in the same order as the score rows.
student_averages = np.mean(scores, axis=1)  # Calculate one average for each student row.
quiz_averages = np.mean(scores, axis=0)  # Calculate one average for each quiz column.
student_highs = np.max(scores, axis=1)  # Find the highest quiz score for each student row.
student_spread = np.std(scores, axis=1)  # Measure each student's score variation across quizzes.
top_student_index = np.argmax(student_averages)  # Find the row index with the highest student average.
top_student_name = student_names[top_student_index]  # Select the name at the highest-average row index.
print("Student averages:", np.round(student_averages, 2))  # Display student averages rounded to two decimals.
print("Quiz averages:", np.round(quiz_averages, 2))  # Display quiz averages rounded to two decimals.
print("Student high scores:", student_highs)  # Display each student's highest score.
print("Student standard deviations:", np.round(student_spread, 2))  # Display each student's variation rounded to two decimals.
print("Highest average:", top_student_name)  # Display the name of the highest-average student.
```

## 7. Generate reproducible practice data

Create a reproducible random-number generator using seed `2026`. Generate a 6-by-5 array of integer game scores from 50 through 100, inclusive. Calculate the overall mean.

**Hint:** Use `np.random.default_rng(2026)`. The upper limit in `.integers(...)` is excluded.

```python
rng = np.random.default_rng(2026)  # Create a reproducible random-number generator with seed 2026.
random_game_scores = rng.integers(50, 101, size=(6, 5))  # Generate six players by five rounds with scores from 50 through 100.
random_overall_mean = np.mean(random_game_scores)  # Calculate one mean using all generated game scores.
print(random_game_scores)  # Display the reproducible random game-score array.
print("Overall mean:", np.round(random_overall_mean, 2))  # Display the mean rounded to two decimals.
```

## 8. Mini-project: Student Score Investigator

Use `student_names` and `scores` to answer these questions with NumPy:

1. Who improved the most from Quiz 1 to Quiz 4?
2. Which quiz had the highest class average?
3. Who was most consistent across all four quizzes?
4. Which students earned an average of at least 85?

Print each answer with a helpful label.

**Hints:**

- Improvement = last column minus first column.
- Add `1` to a zero-based quiz index when displaying a human-friendly quiz number.
- Smaller standard deviation means more consistent.
- A Boolean mask can filter the names array.

```python
improvements = scores[:, -1] - scores[:, 0]  # Calculate each student's change from the first quiz to the last quiz.
most_improved_index = np.argmax(improvements)  # Find the row index with the largest improvement.
most_improved_name = student_names[most_improved_index]  # Select the name of the most improved student.
quiz_averages = np.mean(scores, axis=0)  # Calculate one class average for each quiz column.
best_quiz_number = np.argmax(quiz_averages) + 1  # Convert the highest-average quiz index into a human-friendly quiz number.
student_spread = np.std(scores, axis=1)  # Calculate score variation for each student row.
most_consistent_index = np.argmin(student_spread)  # Find the row index with the smallest standard deviation.
most_consistent_name = student_names[most_consistent_index]  # Select the name of the most consistent student.
student_averages = np.mean(scores, axis=1)  # Calculate one average for each student row.
students_at_least_85 = student_names[student_averages >= 85]  # Filter names using the at-least-85 average mask.
print("Most improved:", most_improved_name, "with", improvements[most_improved_index], "points")  # Display the greatest improvement with its size.
print("Highest class average: Quiz", best_quiz_number)  # Display the quiz number with the highest class average.
print("Most consistent:", most_consistent_name)  # Display the student with the smallest score variation.
print("Average of at least 85:", students_at_least_85)  # Display names of students meeting the average threshold.
```

## Written interpretation - sample answers

1. **Most consistent:** Bella has the smallest standard deviation, so her four scores vary the least around her own average.
2. **Average versus improvement:** No. Bella has the highest average, while Chloe improves the most from Quiz 1 to Quiz 4.
3. **Possibly easiest quiz:** Quiz 4 has the highest class average. It may have been easiest, but the average alone cannot prove difficulty because preparation, topic familiarity, or teaching could also affect performance.
4. **Limitation:** Four quiz scores do not measure all learning, effort, growth, creativity, test conditions, or differences in quiz difficulty.
