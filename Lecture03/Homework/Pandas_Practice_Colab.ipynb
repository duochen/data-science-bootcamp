## 1. Getting Started

```python
# Import the pandas library and give it the short nickname 'pd' (the standard)
import pandas as pd
# Import numpy too — pandas is built on top of it
import numpy as np
```

## 2. Pandas Series

A **Series** is a single column of data with a label (index) for each value — like one column of a spreadsheet.

```python
# Create a Series from a Python list
# On the left you see the automatic index (0, 1, 2, ...) and on the right the values
my_series = pd.Series([10, 20, 30, 40, 50])
print(my_series)
```

```python
# We can give the values custom labels using the 'index' parameter
my_series = pd.Series([10, 20, 30, 40, 50],
                      index=["num1", "num2", "num3", "num4", "num5"])
print(my_series)
```

```python
# A Series can also be built from a Python dictionary
# The dictionary KEYS become the index, the VALUES become the data
my_dict = {'num1': 6, 'num2': 7, 'num3': 8}
my_series = pd.Series(my_dict)
print(my_series)
```

```python
# Access a Series item by its label using square brackets
my_series = pd.Series([10, 20, 30, 40, 50],
                      index=["num1", "num2", "num3", "num4", "num5"])
# Get the value at label 'num3' (returns 30)
print(my_series["num3"])
# Access by POSITION using .iloc — position 0 is the first item
print(my_series.iloc[0])
```

```python
# Pandas Series have built-in statistics methods
my_series = pd.Series([5, 8, 2, 11, 9])
# The smallest value in the series
print("min:", my_series.min())
# The largest value in the series
print("max:", my_series.max())
# The average value
print("mean:", my_series.mean())
# The middle value when sorted
print("median:", my_series.median())
```

## 3. Pandas DataFrame

A **DataFrame** is a full table with rows and columns — think of it as a collection of Series that share the same index.

```python
# Build a DataFrame from a dictionary
# Each KEY becomes a column name, and each list of VALUES becomes the column data
scores = {
    'Subject': ["Mathematics", "History", "English", "Science", "Arts"],
    'Score':   [98, 75, 68, 82, 99]
}
# Convert the dictionary into a DataFrame
my_df = pd.DataFrame(scores)
# Print shows plain text, but returning the DataFrame at the end of a cell shows a pretty table
my_df
```

```python
# You can also build a DataFrame from a list of dictionaries — one dict per row
# If a key is missing from a row, Pandas fills it with NaN (Not a Number = missing value)
scores = [
    {'Subject': 'Mathematics', 'Score': 85},
    {'Subject': 'History',     'Score': 98},
    {'Subject': 'English',     'Score': 76},
    {'Score': 72},              # 'Subject' is missing → will show NaN
    {'Subject': 'Arts',        'Score': 95},
]
my_df = pd.DataFrame(scores)
my_df
```

## 4. Exploring a DataFrame

Whenever you get a new dataset, these four methods are your best friends.

```python
# .head(N) shows the FIRST N rows (default 5) — great for a quick preview
my_df.head(3)
```

```python
# .tail(N) shows the LAST N rows
my_df.tail(2)
```

```python
# .info() prints a summary: number of rows, column names, data types, and how many non-null values
my_df.info()
```

```python
# .describe() prints statistics for every NUMERIC column: count, mean, std, min, quartiles, max
my_df.describe()
```

## 6. The Titanic Dataset

Seaborn is another data-science library that ships with several practice datasets, including the famous **Titanic passenger list**. We'll use it for the rest of the notebook.

```python
# Import seaborn (already installed in Colab) to load the built-in Titanic dataset
import seaborn as sns
# Load the Titanic dataset — each row is a passenger
titanic_data = sns.load_dataset('titanic')
# Preview the first 5 passengers
titanic_data.head()
```

```python
# How big is the dataset? .shape returns (rows, columns)
print(titanic_data.shape)
```

## 7. Selecting Columns with Square Brackets `[]`

```python
# Pass ONE column name in square brackets to get that column back as a Series
age_column = titanic_data["age"]
# Check the type — it is a Series (one column)
print(type(age_column))
# Show the first few values
age_column.head()
```

```python
# Pass a LIST of column names to get a smaller DataFrame with just those columns
subset = titanic_data[["class", "sex", "age"]]
subset.head()
```

## 8. Filtering Rows by a Condition

We write a condition inside `[]`, and pandas keeps only the rows where the condition is `True`.

```python
# Keep only rows where the 'sex' column equals "male"
males = titanic_data[titanic_data["sex"] == "male"]
males.head()
```

```python
# We can combine two conditions with '&' (AND) — remember to wrap each condition in ()
# Keep rows where the passenger is male AND traveled in First class
first_class_males = titanic_data[
    (titanic_data["sex"] == "male") &
    (titanic_data["class"] == "First")
]
first_class_males.head()
```

```python
# .isin(list) checks whether each value is inside a given list — useful for "any of these" filters
ages = [20, 21, 22]
# Keep only passengers aged 20, 21, or 22
young = titanic_data[titanic_data["age"].isin(ages)]
young.head()
```

## 9. Selecting Rows and Columns with `.loc` (by LABEL)

`loc` uses the row's LABEL (its index name) and the column's NAME.

```python
# Build a small student DataFrame with CUSTOM row labels ('Student1' … 'Student5')
scores = [
    {'Subject': 'Mathematics', 'Score': 85, 'Grade': 'B', 'Remarks': 'Good'},
    {'Subject': 'History',     'Score': 98, 'Grade': 'A', 'Remarks': 'Excellent'},
    {'Subject': 'English',     'Score': 76, 'Grade': 'C', 'Remarks': 'Fair'},
    {'Subject': 'Science',     'Score': 72, 'Grade': 'C', 'Remarks': 'Fair'},
    {'Subject': 'Arts',        'Score': 95, 'Grade': 'A', 'Remarks': 'Excellent'},
]
# 'index=' assigns each row a custom label instead of the default 0..4
my_df = pd.DataFrame(scores, index=["Student1", "Student2", "Student3", "Student4", "Student5"])
my_df
```

```python
# Get one row by its label — returns a Series
my_df.loc["Student1"]
```

```python
# Slice from one label to another — with loc BOTH ends are INCLUDED
my_df.loc["Student1":"Student3"]
```

```python
# loc takes two things: [row labels, column names]
# Rows from Student1 to Student4, only the 'Grade' and 'Score' columns
my_df.loc["Student1":"Student4", ["Grade", "Score"]]
```

```python
# loc also works with a boolean condition — great for filtering
# Keep students whose score is greater than 80, and only show Score and Grade
my_df.loc[my_df["Score"] > 80, ["Score", "Grade"]]
```

## 10. Selecting with `.iloc` (by POSITION)

`iloc` uses **integer positions** (0, 1, 2, …) just like Python lists — it ignores the labels.

```python
# Get the row at position 3 (which is the 4th row — remember, positions start at 0)
my_df.iloc[3]
```

```python
# Slice positions 2 to 4 (excluding 4) and columns 0 to 2 (excluding 2)
# With iloc, the END number is NOT included (opposite of loc)
my_df.iloc[2:4, 0:2]
```

## 11. Dropping Rows and Columns

`drop()` returns a **new** DataFrame without the rows or columns you name.

```python
# Rebuild the score DataFrame with the default integer index (0..4)
my_df = pd.DataFrame(scores)
# Drop rows at index 1 and 4 (axis=0 means 'rows')
smaller = my_df.drop([1, 4], axis=0)
smaller
```

```python
# Drop entire columns by name — axis=1 means 'columns'
no_extras = my_df.drop(["Subject", "Grade"], axis=1)
no_extras
```

## 12. Sorting a DataFrame

```python
# Sort passengers from youngest to oldest (ascending is the default)
by_age = titanic_data.sort_values(by="age")
by_age.head()
```

```python
# Sort passengers from OLDEST to youngest by setting ascending=False
by_age_desc = titanic_data.sort_values(by="age", ascending=False)
by_age_desc.head()
```

```python
# Sort by TWO columns: first by 'class', then by 'fare' as a tie-breaker
multi_sort = titanic_data.sort_values(by=["class", "fare"], ascending=[True, False])
multi_sort.head()
```

## 13. Unique Values and Counts

These answer questions like *"how many different classes were there?"* and *"how many passengers were in each class?"*

```python
# .unique() returns the list of DIFFERENT values in a column
titanic_data["class"].unique()
```

```python
# .nunique() returns just the COUNT of different values
titanic_data["class"].nunique()
```

```python
# .value_counts() gives us HOW MANY times each unique value appears
# The output is sorted from most common to least common
titanic_data["class"].value_counts()
```

```python
# .count() returns the number of NON-NULL values in each column
# Columns with lots of missing data (like 'deck') will show smaller numbers
titanic_data.count()
```

## 14. Bonus — A Quick Chart

Pandas has plotting built in, so we can chart our answers instantly.

```python
# Get the passenger counts by class and draw them as a bar chart
# kind='bar' tells pandas to draw bars; the title labels the chart
titanic_data["class"].value_counts().plot(kind="bar", title="Passengers per Class")
```
